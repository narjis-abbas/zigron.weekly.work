```python
class bankaccount:

    def __init__(self, name, acc_no, balance):
        self.name = name
        self.acc_no = acc_no
        self.__balance = balance

    def deposit(self, amount):
        if amount < 1000 or amount > 100000:
            print("Amount must be between 1000 to 100000")
        else:
            self.__balance += amount
            print(f"Amount {amount} deposited successfully")

    def withdraw(self, amount):
        if amount > self.__balance:
            print("Insufficient balance!")
        else:
            self.__balance -= amount
            print(f"Amount {amount} withdrawn successfully")

    def check_balance(self):
        print("Current balance:", self.__balance)

    def show_details(self):
        return f"Name: {self.name}, Account_no: {self.acc_no}, balance: {self.__balance}"


class savingaccount(bankaccount):

    def __init__(self, name, acc_no, balance, interest=0.10):
        super().__init__(name, acc_no, balance)
        self.__interest = interest

    def interest(self):
        return f"Interest of {self._bankaccount__balance * self.__interest/100} is added!"


ali = bankaccount("Ali", 101, 5000)

print(ali.show_details())
ali.deposit(800)
ali.withdraw(2000)
ali.check_balance()


saad = savingaccount("Saad", 102, 5000)

print(saad.show_details())
saad.deposit(1600)
print(saad.interest())
saad.check_balance()
```
